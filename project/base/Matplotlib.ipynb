## 1. Gráficos Simples

```python
import matplotlib.pyplot as plt

# Dados de exemplo
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

# Criar um gráfico de linha
plt.plot(x, y)
plt.title('Gráfico Simples')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.show()
```

## 2. Customização de Gráficos

```python
# Customizar cores, estilos e marcadores
plt.plot(x, y, color='red', linestyle='--', marker='o')
plt.title('Gráfico Customizado')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.grid(True)
plt.show()
```

## 3. Gráficos de Barras

```python
# Dados de exemplo
categorias = ['A', 'B', 'C', 'D']
valores = [3, 7, 8, 5]

# Criar um gráfico de barras
plt.bar(categorias, valores, color='blue')
plt.title('Gráfico de Barras')
plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.show()
```

## 4. Gráficos de Dispersão

```python
# Dados de exemplo
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]

# Criar um gráfico de dispersão
plt.scatter(x, y, color='green')
plt.title('Gráfico de Dispersão')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.show()
```

## 5. Subplots (Vários Gráficos na Mesma Figura)

```python
# Criar múltiplos subplots
plt.figure(figsize=(10, 5))

# Primeiro gráfico
plt.subplot(1, 2, 1)
plt.plot(x, y, color='red')
plt.title('Gráfico 1')

# Segundo gráfico
plt.subplot(1, 2, 2)
plt.bar(categorias, valores, color='blue')
plt.title('Gráfico 2')

plt.tight_layout()
plt.show()
```

## 6. Gráficos de Pizza

```python
# Dados de exemplo
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]

# Criar um gráfico de pizza
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Gráfico de Pizza')
plt.show()
```

## 7. Histogramas

```python
# Dados de exemplo
dados = [22, 87, 5, 43, 56, 73, 55, 54, 11, 20, 51, 5, 79, 31, 27]

# Criar um histograma
plt.hist(dados, bins=5, color='purple', alpha=0.7)
plt.title('Histograma')
plt.xlabel('Intervalos')
plt.ylabel('Frequência')
plt.show()
```

## 8. Gráficos 3D

```python
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Criar dados
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x = np.random.rand(50)
y = np.random.rand(50)
z = np.random.rand(50)

# Gráfico 3D
ax.scatter(x, y, z, c='r', marker='o')
ax.set_xlabel('Eixo X')
ax.set_ylabel('Eixo Y')
ax.set_zlabel('Eixo Z')
plt.title('Gráfico 3D')
plt.show()
```
