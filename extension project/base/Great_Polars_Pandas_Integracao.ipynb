# Integração de Pandas com Great Expectations e Polars
Este notebook demonstra como utilizar as bibliotecas **Pandas**, **Great Expectations** e **Polars** juntas.

## 1. Validação de Dados com Great Expectations e Pandas

```python
import pandas as pd
import great_expectations as ge

# Criar um DataFrame com pandas
df = pd.DataFrame({
    'nome': ['Alice', 'Bob', None],
    'idade': [25, 30, 35]
})

# Converter para um DataFrame do Great Expectations
ge_df = ge.from_pandas(df)

# Validar que a coluna 'nome' não possui valores nulos
expectation = ge_df.expect_column_values_to_not_be_null('nome')

# Resultado da validação
print("Resultado da validação para 'nome':", expectation['success'])

# Validar que a coluna 'idade' está entre 20 e 40
expectation = ge_df.expect_column_values_to_be_between(
    column='idade', min_value=20, max_value=40
)
print("Resultado da validação para 'idade':", expectation['success'])
```

## 2. Manipulação de Dados com Polars e Pandas

```python
import polars as pl

# Criar um DataFrame com pandas
df_pandas = pd.DataFrame({
    'nome': ['Alice', 'Bob', 'Charlie'],
    'idade': [25, 30, 35]
})

# Converter para Polars
df_polars = pl.from_pandas(df_pandas)

# Filtrar dados com Polars
df_polars_filtered = df_polars.filter(pl.col('idade') > 30)
print("DataFrame filtrado com Polars:")
print(df_polars_filtered)

# Converter de volta para Pandas
df_pandas_final = df_polars_filtered.to_pandas()
print("DataFrame convertido de volta para Pandas:")
print(df_pandas_final)
```

## 3. Integração entre Polars e Great Expectations via Pandas

```python
# Converter um DataFrame de Polars para Pandas
df_pandas_from_polars = df_polars.to_pandas()

# Validar com Great Expectations
ge_df = ge.from_pandas(df_pandas_from_polars)
expectation = ge_df.expect_column_to_exist('nome')
print("A coluna 'nome' existe no DataFrame:", expectation['success'])
```
