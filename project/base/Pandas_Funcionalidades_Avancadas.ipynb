## 1. Agrupamento de Dados

```python
import pandas as pd

# Criando um DataFrame de exemplo
df = pd.DataFrame({
    'Categoria': ['A', 'B', 'A', 'B'],
    'Valor': [10, 20, 15, 25]
})

# Agrupando por categoria e calculando a média
agrupado = df.groupby('Categoria')['Valor'].mean()
print(agrupado)
```

## 2. Manipulação de Datas e Horas

```python
# Criando um DataFrame com datas
df = pd.DataFrame({
    'Data': ['2024-01-01', '2024-02-01', '2024-03-01'],
    'Valor': [100, 200, 300]
})

# Convertendo para datetime
df['Data'] = pd.to_datetime(df['Data'])

# Extraindo ano e mês
df['Ano'] = df['Data'].dt.year
df['Mes'] = df['Data'].dt.month
print(df)
```

## 3. Pivot Tables

```python
# Criando uma tabela dinâmica
df = pd.DataFrame({
    'Categoria': ['A', 'B', 'A', 'B'],
    'Ano': [2023, 2023, 2024, 2024],
    'Valor': [10, 20, 15, 25]
})

tabela_pivot = pd.pivot_table(
    df, 
    values='Valor', 
    index='Categoria', 
    columns='Ano', 
    aggfunc='sum', 
    fill_value=0
)
print(tabela_pivot)
```

## 4. Operações com Dados Faltantes

```python
# Criando um DataFrame com valores ausentes
df = pd.DataFrame({
    'A': [1, None, 3],
    'B': [4, 5, None]
})

# Preenchendo valores nulos com a média
df['A'] = df['A'].fillna(df['A'].mean())
print(df)
```

## 5. Merge e Join

```python
# Criando dois DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nome': ['Alice', 'Bob', 'Charlie']
})
df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Idade': [25, 30, 35]
})

# Fazendo merge
df_combinado = pd.merge(df1, df2, on='ID', how='inner')
print(df_combinado)
```

## 6. Aplicação de Funções Personalizadas

```python
# Criando um DataFrame
df = pd.DataFrame({
    'Valor': [1, 2, 3, 4]
})

# Aplicando uma função personalizada
df['Quadrado'] = df['Valor'].apply(lambda x: x**2)
print(df)
```
