# Limpeza de Dados em Python
Este notebook apresenta ferramentas e exemplos práticos para realizar a limpeza de dados em Python.

## 1. Limpeza com Pandas

```python
import pandas as pd

# Exemplo de DataFrame
df = pd.DataFrame({
    'Coluna1': [1, None, 3, None],
    'Coluna2': ['a', 'b', None, 'd'],
    'Coluna3': [10, 20, 30, 40]
})

# Remover valores ausentes
df.dropna(inplace=True)
print("Após remoção de valores nulos:")
print(df)

# Preencher valores ausentes com um valor específico
df = pd.DataFrame({
    'Coluna1': [1, None, 3, None],
    'Coluna2': ['a', 'b', None, 'd'],
    'Coluna3': [10, 20, 30, 40]
})
df.fillna(0, inplace=True)
print("Após preenchimento de valores nulos com 0:")
print(df)

# Remoção de duplicatas
df = pd.DataFrame({
    'Coluna1': [1, 2, 2, 3],
    'Coluna2': ['a', 'b', 'b', 'd']
})
df.drop_duplicates(inplace=True)
print("Após remoção de duplicatas:")
print(df)
```

## 2. Limpeza com NumPy

```python
import numpy as np

# Exemplo de array
arr = np.array([1, -1, np.nan, 4, -5])

# Substituir valores negativos por 0
arr[arr < 0] = 0
print("Array após substituição de valores negativos por 0:", arr)

# Remover valores NaN
arr = arr[~np.isnan(arr)]
print("Array após remoção de valores NaN:", arr)
```

## 3. Limpeza com Regex

```python
import re

# Exemplo de texto
texto = "Texto com caracteres indesejados! @#123"

# Remover caracteres especiais
texto_limpo = re.sub(r'[^a-zA-Z0-9\s]', '', texto)
print("Texto após remoção de caracteres especiais:", texto_limpo)

# Extrair números do texto
numeros = re.findall(r'\d+', texto)
print("Números extraídos do texto:", numeros)
```

## 4. Detectar Outliers com SciPy

```python
from scipy.stats import zscore

# Exemplo de DataFrame
df = pd.DataFrame({
    'Valores': [10, 12, 14, 100, 12, 13, 15]
})

# Calcular z-score
df['z_score'] = zscore(df['Valores'])

# Filtrar valores com z-score menor que 3
df_filtrado = df[df['z_score'] < 3]
print("DataFrame após remoção de outliers:")
print(df_filtrado)
```

## 5. Limpeza Textual com NLTK

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

# Exemplo de texto
texto = "Este é um exemplo de texto com palavras irrelevantes."

# Remover stopwords
stopwords_portugues = stopwords.words('portuguese')
palavras = [word for word in texto.split() if word.lower() not in stopwords_portugues]
print("Texto após remoção de stopwords:", ' '.join(palavras))
```
