# Public Data Sources
Explore various public data sources such as government databases, open data platforms, and research datasets.

```python
# Public Data Sources

# Explore various public data sources such as government databases, open data platforms, and research datasets.

# Import necessary libraries
import pandas as pd

# Example 1: Government Database - US Census Data
census_data_url = "https://api.census.gov/data/2019/pep/population?get=NAME,POP&for=state:*"
census_data = pd.read_json(census_data_url)
census_data.columns = census_data.iloc[0]
census_data = census_data[1:]
census_data.head()

# Example 2: Open Data Platform - World Bank Data
world_bank_data_url = "http://api.worldbank.org/v2/country/all/indicator/SP.POP.TOTL?format=json&date=2019"
world_bank_data = pd.read_json(world_bank_data_url)
world_bank_data = pd.json_normalize(world_bank_data[1])
world_bank_data.head()

# Example 3: Research Dataset - UCI Machine Learning Repository
uci_data_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
uci_column_names = ["sepal_length", "sepal_width", "petal_length", "petal_width", "class"]
uci_data = pd.read_csv(uci_data_url, header=None, names=uci_column_names)
uci_data.head()
```

# APIs for Data Retrieval
Demonstrate how to use APIs to retrieve data from various online services and platforms.

```python
# APIs for Data Retrieval

# Import necessary libraries
import requests

# Example 1: RESTful API - OpenWeatherMap API
api_key = "your_openweathermap_api_key"
city = "London"
weather_url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}"
weather_response = requests.get(weather_url)
weather_data = weather_response.json()

# Extract relevant information
weather_info = {
    "city": weather_data["name"],
    "temperature": weather_data["main"]["temp"],
    "weather": weather_data["weather"][0]["description"]
}
weather_info_df = pd.DataFrame([weather_info])
weather_info_df.head()

# Example 2: RESTful API - GitHub API
github_user = "octocat"
github_url = f"https://api.github.com/users/{github_user}/repos"
github_response = requests.get(github_url)
github_data = github_response.json()

# Extract relevant information
github_repos = []
for repo in github_data:
    github_repos.append({
        "name": repo["name"],
        "url": repo["html_url"],
        "created_at": repo["created_at"]
    })
github_repos_df = pd.DataFrame(github_repos)
github_repos_df.head()

# Example 3: GraphQL API - SpaceX API
spacex_url = "https://api.spacex.land/graphql/"
query = """
{
  launchesPast(limit: 3) {
    mission_name
    launch_date_utc
    rocket {
      rocket_name
    }
  }
}
"""
spacex_response = requests.post(spacex_url, json={'query': query})
spacex_data = spacex_response.json()

# Extract relevant information
spacex_launches = []
for launch in spacex_data["data"]["launchesPast"]:
    spacex_launches.append({
        "mission_name": launch["mission_name"],
        "launch_date": launch["launch_date_utc"],
        "rocket_name": launch["rocket"]["rocket_name"]
    })
spacex_launches_df = pd.DataFrame(spacex_launches)
spacex_launches_df.head()
```

# Web Scraping Techniques
Introduce web scraping techniques to extract data from websites, including the use of libraries like BeautifulSoup and Scrapy.

```python
# Web Scraping Techniques

# Import necessary libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Example 1: Using BeautifulSoup to scrape data from a website
url = "https://example.com"
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')

# Extract data from the website
titles = soup.find_all('h2')
titles_text = [title.get_text() for title in titles]

# Create a DataFrame
titles_df = pd.DataFrame(titles_text, columns=['Title'])
titles_df.head()

# Example 2: Using Scrapy to scrape data from a website
# Note: Scrapy is typically run from the command line, but here is an example of a Scrapy spider

import scrapy
from scrapy.crawler import CrawlerProcess

class ExampleSpider(scrapy.Spider):
    name = "example"
    start_urls = ['https://example.com']

    def parse(self, response):
        for title in response.css('h2::text').getall():
            yield {'Title': title}

# Run the Scrapy spider
process = CrawlerProcess(settings={
    "FEEDS": {
        "titles.json": {"format": "json"},
    },
})

process.crawl(ExampleSpider)
process.start()

# Load the scraped data into a DataFrame
scrapy_data = pd.read_json('titles.json')
scrapy_data.head()
```
