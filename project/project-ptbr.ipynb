# Análise das Instituições de Educação Superior (IES)
Analisar o número e tipos de instituições (públicas ou privadas) e suas classificações acadêmicas.

```python
# Análise das Instituições de Educação Superior (IES)

# Filtrar dados relevantes para análise das IES
ies_dados = microdados_ies[['NO_REGIAO_IES', 'TP_REDE', 'TP_ORGANIZACAO_ACADEMICA']]

# Renomear colunas para melhor compreensão
ies_dados.columns = ['Região', 'Rede', 'Organização Acadêmica']

# Mapear valores numéricos para categorias
rede_map = {1: 'Pública', 2: 'Privada'}
org_acad_map = {
    1: 'Universidade',
    2: 'Centro Universitário',
    3: 'Faculdade',
    4: 'Instituto Federal de Educação, Ciência e Tecnologia',
    5: 'Centro Federal de Educação Tecnológica'
}

ies_dados['Rede'] = ies_dados['Rede'].map(rede_map)
ies_dados['Organização Acadêmica'] = ies_dados['Organização Acadêmica'].map(org_acad_map)

# Contar o número de IES por tipo de rede e organização acadêmica
contagem_rede = ies_dados['Rede'].value_counts()
contagem_org_acad = ies_dados['Organização Acadêmica'].value_counts()

# Exibir contagem de IES por tipo de rede
print("\nContagem de IES por Tipo de Rede:")
print(contagem_rede)

# Exibir contagem de IES por tipo de organização acadêmica
print("\nContagem de IES por Tipo de Organização Acadêmica:")
print(contagem_org_acad)

# Visualizar a distribuição das IES por tipo de rede
plt.figure(figsize=(10, 6))
sns.countplot(data=ies_dados, x='Rede', palette='viridis')
plt.title('Distribuição das IES por Tipo de Rede')
plt.xlabel('Tipo de Rede')
plt.ylabel('Quantidade de IES')
plt.show()

# Visualizar a distribuição das IES por tipo de organização acadêmica
plt.figure(figsize=(12, 8))
sns.countplot(data=ies_dados, y='Organização Acadêmica', palette='viridis', order=contagem_org_acad.index)
plt.title('Distribuição das IES por Tipo de Organização Acadêmica')
plt.xlabel('Quantidade de IES')
plt.ylabel('Tipo de Organização Acadêmica')
plt.show()

# Analisar a distribuição das IES por região e tipo de rede
plt.figure(figsize=(14, 8))
sns.countplot(data=ies_dados, x='Região', hue='Rede', palette='viridis')
plt.title('Distribuição das IES por Região e Tipo de Rede')
plt.xlabel('Região')
plt.ylabel('Quantidade de IES')
plt.legend(title='Tipo de Rede')
plt.show()

# Analisar a distribuição das IES por região e tipo de organização acadêmica
plt.figure(figsize=(14, 8))
sns.countplot(data=ies_dados, x='Região', hue='Organização Acadêmica', palette='viridis')
plt.title('Distribuição das IES por Região e Tipo de Organização Acadêmica')
plt.xlabel('Região')
plt.ylabel('Quantidade de IES')
plt.legend(title='Tipo de Organização Acadêmica')
plt.show()
```

# Desenvolvimento do Questionário de Afinidade Vocacional
Desenvolver um questionário que correlacione respostas dos estudantes às áreas de cursos de graduação.

```python
# Desenvolvimento do Questionário de Afinidade Vocacional

# Definir perguntas do questionário
perguntas = [
    "Você prefere atividades práticas ou teóricas?",
    "Você gosta de trabalhar em equipe?",
    "Você prefere trabalhar em ambientes internos ou externos?",
    "Você se interessa por tecnologia?",
    "Você gosta de resolver problemas complexos?",
    "Você prefere atividades criativas ou analíticas?",
    "Você se interessa por ciências humanas, exatas ou biológicas?",
    "Você gosta de atividades que envolvem comunicação?",
    "Você prefere trabalhar com números ou com pessoas?",
    "Você se interessa por empreendedorismo?"
]

# Definir áreas de conhecimento e suas características
areas_conhecimento = {
    "Engenharia": ["práticas", "tecnologia", "resolver problemas complexos", "analíticas", "ciências exatas", "números"],
    "Ciências Humanas": ["teóricas", "trabalhar em equipe", "ciências humanas", "comunicação", "pessoas"],
    "Ciências Biológicas": ["práticas", "ambientes externos", "ciências biológicas", "resolver problemas complexos"],
    "Tecnologia da Informação": ["tecnologia", "resolver problemas complexos", "analíticas", "ciências exatas", "números"],
    "Artes": ["criativas", "ciências humanas", "comunicação", "atividades criativas"],
    "Administração": ["trabalhar em equipe", "empreendedorismo", "ciências humanas", "comunicação", "pessoas"]
}

# Função para calcular afinidade com áreas de conhecimento
def calcular_afinidade(respostas):
    afinidade = {area: 0 for area in areas_conhecimento}
    for i, resposta in enumerate(respostas):
        for area, caracteristicas in areas_conhecimento.items():
            if resposta in caracteristicas:
                afinidade[area] += 1
    return afinidade

# Exemplo de respostas de um estudante
respostas_estudante = [
    "práticas", "trabalhar em equipe", "ambientes internos", "tecnologia", 
    "resolver problemas complexos", "analíticas", "ciências exatas", 
    "comunicação", "números", "empreendedorismo"
]

# Calcular afinidade do estudante com as áreas de conhecimento
afinidade_estudante = calcular_afinidade(respostas_estudante)

# Exibir afinidade do estudante com as áreas de conhecimento
print("\nAfinidade do Estudante com as Áreas de Conhecimento:")
for area, afinidade in afinidade_estudante.items():
    print(f"{area}: {afinidade}")

# Visualizar afinidade do estudante com as áreas de conhecimento
plt.figure(figsize=(12, 8))
sns.barplot(x=list(afinidade_estudante.keys()), y=list(afinidade_estudante.values()), palette='viridis')
plt.title('Afinidade do Estudante com as Áreas de Conhecimento')
plt.xlabel('Área de Conhecimento')
plt.ylabel('Afinidade')
plt.show()
```
