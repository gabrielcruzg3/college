## 1. Importar as bibliotecas necessárias

```python
import pandas as pd
```

## 2. Ler o arquivo Excel

```python
# Substitua 'seu_arquivo.xlsx' pelo caminho do seu arquivo
df = pd.read_excel('seu_arquivo.xlsx')
```

## 3. Explorar os dados

```python
# Mostra as primeiras 5 linhas
print(df.head())

# Mostra informações sobre as colunas e tipos de dados
print(df.info())

# Exibe estatísticas descritivas
print(df.describe())
```

## 4. Manipular os dados

```python
# Selecionar colunas específicas
coluna_especifica = df['Nome_da_Coluna']
multiplas_colunas = df[['Coluna1', 'Coluna2']]

# Filtrar linhas
filtro = df[df['Idade'] > 18]

# Renomear colunas
df.rename(columns={'Coluna_Antiga': 'Coluna_Nova'}, inplace=True)

# Adicionar ou modificar colunas
df['Nova_Coluna'] = df['Coluna_Existente'] * 2

# Remover colunas ou linhas
df.drop(columns=['Coluna1'], inplace=True)
df.drop(index=[0, 1], inplace=True)

# Lidar com valores ausentes
print(df.isnull().sum())
df.fillna(0, inplace=True)
df.dropna(inplace=True)
```

## 5. Ordenar os dados

```python
# Ordenar por uma coluna
df.sort_values(by='Idade', ascending=True, inplace=True)
```

## 6. Salvar o arquivo Excel tratado

```python
# Salvar os dados tratados em um novo arquivo
df.to_excel('arquivo_tratado.xlsx', index=False)
```

## Exemplo completo

```python
import pandas as pd

# Carregar o arquivo Excel
df = pd.read_excel('dados.xlsx')

# Visualizar informações básicas
print(df.head())
print(df.info())

# Filtrar dados
df_filtrado = df[df['Idade'] > 18]

# Adicionar nova coluna
df_filtrado['Idade_Dobrada'] = df_filtrado['Idade'] * 2

# Remover valores nulos
df_filtrado.dropna(inplace=True)

# Salvar os dados tratados
df_filtrado.to_excel('dados_tratados.xlsx', index=False)
```
